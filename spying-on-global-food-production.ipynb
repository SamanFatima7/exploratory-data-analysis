1. **Net Food Importing Developing Countries (FAO):**
   Countries that rely on importing a significant portion of their food due to insufficient domestic production.

2. **Small Island Developing States (FAO):**
   Geopolitical entities characterized by their small land area, limited resources, and vulnerability to environmental challenges, often located in island regions.

3. **Upper-middle-income countries:**
   Countries with a moderate level of economic development, typically measured by gross national income (GNI) per capita.

4. **Lower-middle-income countries:**
   Countries with a moderate but lower level of economic development, typically measured by gross national income (GNI) per capita.

5. **Low Income Food Deficit Countries (FAO):**
   Nations that experience challenges in providing adequate food for their populations due to low income levels and deficits in food production.

6. **High-income countries:**
   Nations with a high level of economic development, often characterized by a high gross national income (GNI) per capita.

7. **Land Locked Developing Countries (FAO):**
   Countries that lack direct access to the sea and may face challenges in international trade due to their landlocked geographical position.

8. **Least Developed Countries (FAO):**
   Nations facing significant economic and social challenges, as identified by the United Nations, including low income, weak human development indicators, and economic vulnerability.

```python
import plotly.express as px
import pandas as pd
import warnings

# Suppress FutureWarning from plotly
warnings.simplefilter(action='ignore', category=FutureWarning)

# Example 1: Line chart for Maize Production over the years
fig_maize = px.line(df, x='Year', y='Maize Production (tonnes)', color='Country_Name', title='Maize Production Over Time')
fig_maize.show()

# Example 2: Bar chart for Rice Production in a specific year
fig_rice = px.bar(df[df['Year'] == 2020], x='Country_Name', y='Rice  Production ( tonnes)', title='Rice Production in 2020')
fig_rice.show()

# Example 3: Scatter plot for Wheat Production vs. Yams Production
fig_scatter = px.scatter(df, x='Wheat Production (tonnes)', y='Yams  Production (tonnes)', color='Country_Name',
                         title='Wheat vs. Yams Production')
fig_scatter.show()

# Example 4: Stacked area chart for total production of selected crops over the years
selected_crops = ['Maize Production (tonnes)', 'Rice  Production ( tonnes)', 'Wheat Production (tonnes)']

df_selected_crops = df[['Year', 'Country_Name'] + selected_crops]

with pd.option_context('mode.chained_assignment', None):  # Suppress the SettingWithCopyWarning
    grouped = df_selected_crops.groupby(['Year', 'Country_Name'], observed=False)

fig_area = px.area(grouped.sum().reset_index(), x='Year', y=selected_crops, color='Country_Name',
                   title='Total Production of Selected Crops Over Time')
fig_area.show()
```
