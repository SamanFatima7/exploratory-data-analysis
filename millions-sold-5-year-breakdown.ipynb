# <span style='color:#E1B12D;'>Meta Data</span><br>

<i>This dataset comprises a collection of video games with sales exceeding 100,000 copies. It was obtained through scraping <a href="https://www.vgchartz.com/"><b>vgchartz.com</b></a>.</i><br>

<b>Fields include</b><br>

   -  <b>Name</b> - The game's title<br>
   -  <b>Platform</b> - The platform of the game's release (e.g., PC, PS4)<br>
   - <b>Year</b> - The year of the game's release<br>
   - <b>Genre</b> - The genre of the game<br>
   - <b>Publisher</b> - The game's publisher<br>
   - <b>NA_Sales</b> - Sales in North America (in millions)<br>
   - <b>EU_Sales</b> - Sales in Europe (in millions)<br>
   - <b>JP_Sales</b> - Sales in Japan (in millions)<br>
   - <b>Other_Sales</b> - Sales in the rest of the world (in millions)<br>
   - <b>Global_Sales</b> - Total worldwide sales<br>

<i>The script used to scrape the data is available at <a href="https://github.com/GregorUT/vgchartzScrape">GitHub</a>. It is based on BeautifulSoup using Python. There are 16,598 records. Two records were dropped due to incomplete information.</i><br>

<div style="margin-top: 20px; font-family: Arial, sans-serif; font-size: 22px; color: #FFA500; font-weight: bold;">2.3 ||  Dealing With Column [Name]</div>

---



```python
df['Name'].unique
```

```python
#  There is a number of games having their version like Taken 3 , Taken 7 so if i remove digits from it then i will lost information so i have tried this simple but effective way to deal with this 
#  problem in this way i will not lost any information plus point is i would be able to keep it in correct datatype :)  
df['Name'] = df['Name'].str.replace('5', 'Five')
df['Name'] = df['Name'].str.replace('9', 'Nine')
df['Name'] = df['Name'].str.replace('8', 'Eight')
df['Name'] = df['Name'].str.replace('3', 'Three')
df['Name'] = df['Name'].str.replace('2', 'Two')
df['Name'] = df['Name'].str.replace('7', 'Seven')
df['Name'] = df['Name'].str.replace('6', 'Six')
df['Name'] = df['Name'].str.replace('4', 'Four')
df['Name'] = df['Name'].str.replace('1', 'One')
df['Name'] = df['Name'].str.replace('0', 'Zero')
```

```python
# Example: Remove non-alphanumeric characters from the 'Name' column 
df['Name'] = df['Name'].replace('[^a-zA-Z0-9\s]', '', regex=True)
```

```python
# Example: Remove numeric characters from the 'Name' column
df['Name'] = df['Name'].replace('\d', '', regex=True)
```

```python
# Example: Display unique values in the 'Name' column
df['Name']
```

```python
df['Name'] = df['Name'].str.strip()
```

```python
df.dtypes
```

```python
#  printing all unique values to manually check the data from a human eye with eagle eyesight abilities 
unique_names = df['Name'].unique()

# Display unique values using a loop
for name in unique_names:
    print(name)
```
