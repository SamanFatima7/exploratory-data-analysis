<div style="background-color: #f9f9f9; padding: 10px; font-family: Arial, sans-serif;">
    <h1 style="font-family: 'Comic Sans MS', cursive, sans-serif; color: #333;"> Scatter Plots: Discover Relationships and Outliers !</h1>
    <p>Unlocking Insights with Simple Points</p>
    <ul>
        <li><strong>Correlation Check:</strong> See if two variables are related.</li>
        <li><strong>Outlier Detection:</strong> Highlight unusual data points.</li>
        <li><strong>Pattern Spotting:</strong> Reveal any noticeable trends or groupings.</li>
        <li><strong>Data Comparison:</strong> Useful for comparing different datasets or variables.</li>
    </ul>
</div>

```python
import matplotlib.pyplot as plt

# Sample data
x = [1, 3, 5, 7, 9,10,11,34,45,66]
y = [2, 5, 7, 2, 8,2,4,56,76,34]

plt.scatter(x, y)
plt.xlabel("X-axis Label")
plt.ylabel("Y-axis Label")
plt.title("Scatter Plot Example")
plt.show()
```

    <h1 style="font-family: 'Comic Sans MS', cursive, sans-serif; color: #333;">Explore Trends with Line Plots!</h1>

```python
import matplotlib.pyplot as plt

# Sample data (time and temperature)
time = [1, 2, 3, 4, 5]
temperature = [10, 15, 18, 20, 22]

plt.plot(time, temperature)
plt.xlabel("Time")
plt.ylabel("Temperature")
plt.title("Line Plot Example")
plt.show()
```

<div style="background-color: #f9f9f9; padding: 10px; font-family: Arial, sans-serif;">
    <h1 style="font-family: 'Comic Sans MS', cursive, sans-serif; color: #333;">Uncover Data Spread with Histograms!</h1>
    <p>Visualizing Data Distribution</p>
    <ul>
        <li><strong>Data Spread:</strong> Show how data points are distributed across different ranges.</li>
        <li><strong>Pattern Recognition:</strong> Identify common ranges or peaks in data.</li>
        <li><strong>Outlier Detection:</strong> Spot unusual data points.</li>
        <li><strong>Comparison:</strong> Compare distributions of different datasets.</li>
    </ul>
</div>

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data (iris flower petal lengths)
data = np.random.rand(100) * 5  # Simulate some random data

plt.hist(data)
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.title("Histogram Example")
plt.show()
```

<div style="background-color: #f9f9f9; padding: 10px; font-family: Arial, sans-serif;">
    <h1 style="font-family: 'Comic Sans MS', cursive, sans-serif; color: #333;">Compare with Bar Charts!</h1>
    <p>Visualizing Categorical Data</p>
    <ul>
        <li><strong>Comparison:</strong> Easily compare categories or frequencies of data points.</li>
        <li><strong>Categorical Data Representation:</strong> Ideal for representing categorical data or proportions within a dataset.</li>
        <li><strong>Pattern Recognition:</strong> Identify trends or differences in categories.</li>
        <li><strong>Data Distribution:</strong> Understand the distribution of categorical data.</li>
    </ul>
</div>

```python
import matplotlib.pyplot as plt

# Sample data (customer satisfaction ratings)
ratings = ["Excellent", "Good", "Average", "Poor"]
counts = [30, 45, 15, 10]

plt.bar(ratings, counts)
plt.xlabel("Customer Satisfaction Rating")
plt.ylabel("Number of Customers")
plt.title("Bar Chart Example")
plt.show()
```

<div style="background-color: #f9f9f9; padding: 10px; font-family: Arial, sans-serif;">
    <h1 style="font-family: 'Comic Sans MS', cursive, sans-serif; color: #333;">Unveil Insights with Box Plots!</h1>
    <p>Visualizing Data Distribution and Outliers</p>
    <ul>
        <li><strong>Distribution Insight:</strong> Show data distribution within quartiles.</li>
        <li><strong>Outlier Identification:</strong> Highlight outliers in the data.</li>
        <li><strong>Group Comparison:</strong> Compare distributions between different groups.</li>
        <li><strong>Data Skewness Detection:</strong> Identify skewed data points.</li>
    </ul>
</div>

```python
import matplotlib.pyplot as plt

# Sample data (exam scores for two classes)
class1 = [75, 82, 90, 68, 100]
class2 = [85, 93, 78, 88, 95]

plt.boxplot([class1, class2], labels=["Class 1", "Class 2"], notch=True)
plt.ylabel("Exam Scores")
plt.title("Box Plot Example")
plt.show()
```

<div style="background-color: #f9f9f9; padding: 10px; font-family: Arial, sans-serif;">
    <h1 style="font-family: 'Comic Sans MS', cursive, sans-serif; color: #333;">Visualize Proportions with Pie Charts!</h1>
    <p>Highlight Categorical Data Breakdowns</p>
    <ul>
        <li><strong>Proportion Representation:</strong> Represent proportions of a whole dataset.</li>
        <li><strong>Categorical Breakdown:</strong> Showcase categorical data breakdowns.</li>
        <li><strong>Clarity Limitation:</strong> Best for a few categories to maintain clarity.</li>
        <li><strong>Percentage Display:</strong> Easily visualize percentage contributions of each category.</li>
    </ul>
</div>

```python
import matplotlib.pyplot as plt

# Sample data (fruit basket composition)
fruits = ["Apple", "Banana", "Orange", "Grape"]
percentages = [30, 25, 20, 25]

plt.pie(percentages, labels=fruits, autopct="%1.1f%%")
plt.title("Pie Chart Example")
plt.show()
```

<div style="background-color: #f9f9f9; padding: 10px; font-family: Arial, sans-serif;">
    <h1 style="font-family: 'Comic Sans MS', cursive, sans-serif; color: #333;">Explore Relationships with Heatmaps!</h1>
    <p>Visualizing Correlations in Large Datasets</p>
    <ul>
        <li><strong>Relationship Visualization:</strong> Visualize relationships between many variables.</li>
        <li><strong>Color-Coded Matrix:</strong> Represent data as a color-coded matrix.</li>
        <li><strong>Correlation Exploration:</strong> Excellent for exploring correlations within large datasets.</li>

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data (correlation matrix)
correlation_matrix = np.random.rand(10, 10)

plt.matshow(correlation_matrix)
plt.colorbar()
plt.title("Heatmap Example")
plt.show()
```

<div style="background-color: #f9f9f9; padding: 10px; font-family: Arial, sans-serif;">
    <h1 style="font-family: 'Comic Sans MS', cursive, sans-serif; color: #333;">Uncover Insights with Violin Plots!</h1>
    <p>Visualizing Data Distribution and Spread</p>
    <ul>
        <li><strong>Combined Features:</strong> Combine box plots and density plots.</li>
        <li><strong>Distribution and Spread:</strong> Show both distribution and spread of data within categories.</li>
        <li><strong>Category Comparison:</strong> Compare distributions between different categories.</li>
        <li><strong>Outlier Detection:</strong> Identify outliers and unusual data points.</li>

</div>

```python
# (Library: seaborn)
import seaborn as sns
import pandas as pd

# Sample data (iris flower petal lengths by species)
data = sns.load_dataset("iris")
sns.violinplot(x = "species", y = "petal_length", data=data, showmeans=True)
plt.title("Violin Plot Example")
plt.show()
```

<div style="background-color: #f9f9f9; padding: 10px; font-family: Arial, sans-serif;">
    <h1 style="font-family: 'Comic Sans MS', cursive, sans-serif; color: #333;">Track Trends with Area Plots!</h1>
    <p>Visualizing Cumulative Trends over Time</p>
    <ul>
        <li><strong>Cumulative Sum:</strong> Depict the cumulative sum of multiple datasets.</li>
        <li><strong>Trend Visualization:</strong> Useful for visualizing trends over time.</li>
        <li><strong>Proportion Representation:</strong> Show proportions of different datasets.</li>
        <li><strong>Data Comparison:</strong> Compare cumulative trends between datasets.</li>
    </ul>
</div>

```python
import matplotlib.pyplot as plt

# Sample data (sales figures for different product categories)
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales1 = [100, 120, 150, 180, 200]
sales2 = [80, 100, 130, 140, 160]

plt.stackplot(months, sales1, sales2, labels=["Product A", "Product B"])
plt.xlabel("Months")
plt.ylabel("Total Sales")
plt.title("Area Plot Example")
plt.legend()
plt.show()
```
